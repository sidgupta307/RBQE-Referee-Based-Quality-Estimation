# Referee-Based Quality Estimation (RBQE)

Official reproducibility notebook accompanying the paper

This notebook reproduces the principal experimental analyses reported in the manuscript.

import os

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    confusion_matrix
)

from scipy.stats import spearmanr

plt.rcParams["figure.figsize"]=(8,6)

```python
RESULTS_DIR="sample_results"

FIGURE_DIR="figures"
```

```python
scores=pd.read_csv(
    os.path.join(
        RESULTS_DIR,
        "merged_referee_scores.csv"
    )
)

scores.head()
```

```python
scores.describe()
```

## ROC Analysis

failure=(scores["gt_dice"]<0.50).astype(int)

agreement=scores["segformer_score"]

fpr,tpr,_=roc_curve(failure,agreement,pos_label=0)

auc=roc_auc_score(failure,-agreement)

print("ROC-AUC =",auc)

```python
plt.figure()

plt.plot(fpr,tpr,label=f"AUC={auc:.3f}")

plt.plot([0,1],[0,1],'--')

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.legend()

plt.grid(True)

plt.show()
```

## Spearman Correlation

```python
rho,p=spearmanr(
    scores["segformer_score"],
    scores["gt_dice"]
)

print(rho,p)
```
